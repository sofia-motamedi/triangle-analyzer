```python
x1=int(input("please enter x1"))
y1=int(input("please enter y1"))
x2=int(input("please enter x2"))
y2=int(input("please enter y2"))
x3=int(input("please enter x3"))
y3=int(input("please enter y3"))

def distance(x1, y1, x2, y2):
    return ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
a= distance(x1,y1,x2,y2)
b= distance(x2,y2,x3,y3)
c= distance(x3,y3,x1,y1) 

def calculate_area(a,b,c):
    m=(a+b+c)/2
    return (m * (m - a) * (m - b) * (m - c)) ** 0.5
    
if x1>0 and x2>0 and x3>0 and y1>0 and y2>0 and y3>0:
    if a+b>c and a+c>b and b+c>a:
        print ("it can be a triangle")
        d= calculate_area(a,b,c)
        print(d)
        if a == b == c:
            print("Equilateral")
        elif a == b or b == c or c == a:
            print("Isosceles")
        elif a ** 2 + b ** 2 == c ** 2 or b ** 2 + c ** 2 == a ** 2 or c ** 2 + a ** 2 == b ** 2:
            print("Right-angled")
        else:
            print("Scalene")
        
    else:
        print ("it cant be a triangle")
        quit()
else:
    print("please write positive number")
```
